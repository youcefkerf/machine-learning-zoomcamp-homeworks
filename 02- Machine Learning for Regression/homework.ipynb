```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv('car_fuel_efficiency_2026.csv')

df = df[
    [
        'engine_displacement',
        'horsepower',
        'vehicle_weight',
        'model_year',
        'fuel_efficiency_mpg'
    ]
]

print(df.head())
print(df.info())
```

```python
print(df['fuel_efficiency_mpg'].describe())

print(df['fuel_efficiency_mpg'].skew())
```

```python
print(df.isnull().sum())
```

```python
print(df['horsepower'].median())
```

```python
n = len(df)

n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test

np.random.seed(42)

idx = np.arange(n)
np.random.shuffle(idx)

df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train + n_val]]
df_test = df.iloc[idx[n_train + n_val:]]
```

```python
y_train = df_train['fuel_efficiency_mpg'].values
y_val = df_val['fuel_efficiency_mpg'].values

del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
```

```python
def rmse(y, y_pred):
    return np.sqrt(mean_squared_error(y, y_pred))
```

```python
X_train = df_train.fillna(0)
X_val = df_val.fillna(0)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_val)

score = rmse(y_val, y_pred)

print("Q3 - RMSE with 0:", round(score, 3))
```

```python
mean = df_train.mean()

X_train = df_train.fillna(mean)
X_val = df_val.fillna(mean)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_val)

score = rmse(y_val, y_pred)

print("Q3 - RMSE with mean:", round(score, 3))
```

```python
r = [0, 0.01, 0.1, 1, 5, 10, 100]
```

```python
X_train = df_train.fillna(0)
X_val = df_val.fillna(0)
```

```python
for r in [0, 0.01, 0.1, 1, 5, 10, 100]:

    model = Ridge(alpha=r)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_val)

    score = rmse(y_val, y_pred)

    print(r, round(score, 4))
```

```python
scores = []

for seed in range(10):

    n = len(df)

    n_val = int(n * 0.2)
    n_test = int(n * 0.2)
    n_train = n - n_val - n_test

    np.random.seed(seed)

    idx = np.arange(n)
    np.random.shuffle(idx)

    df_train = df.iloc[idx[:n_train]]
    df_val = df.iloc[idx[n_train:n_train + n_val]]

    y_train = df_train['fuel_efficiency_mpg'].values
    y_val = df_val['fuel_efficiency_mpg'].values

    del df_train['fuel_efficiency_mpg']
    del df_val['fuel_efficiency_mpg']

    X_train = df_train.fillna(0)
    X_val = df_val.fillna(0)

    model = LinearRegression()
    model.fit(X_train, y_train)

    y_pred = model.predict(X_val)

    score = rmse(y_val, y_pred)

    scores.append(score)

    print(seed, round(score, 3))
```

```python
std = np.std(scores)

print("Q5 standard deviation:", round(std, 3))
```

```python
n = len(df)

n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test

np.random.seed(9)

idx = np.arange(n)
np.random.shuffle(idx)

df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train + n_val]]
df_test = df.iloc[idx[n_train + n_val:]]
```

```python
df_train_full = pd.concat([df_train, df_val])
```

```python
y_train_full = df_train_full['fuel_efficiency_mpg'].values
y_test = df_test['fuel_efficiency_mpg'].values

X_train_full = df_train_full.drop('fuel_efficiency_mpg', axis=1)
X_test = df_test.drop('fuel_efficiency_mpg', axis=1)
```

```python
X_train_full = X_train_full.fillna(0)
X_test = X_test.fillna(0)
```

```python
model = Ridge(alpha=0.001)

model.fit(X_train_full, y_train_full)

y_pred = model.predict(X_test)

score = rmse(y_test, y_pred)

print("Q6 RMSE:", round(score, 3))
```
